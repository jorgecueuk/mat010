# Teorema Fundamental del Cálculo

## Propiedades

<dl>
<dt>Actividad</dt>
<dd>
$$\int_{a}^{b}f(x)\,dx = \int_{a}^{c}f(x)\,dx + \int_{c}^{b}f(x)\,dx$$
<dt>Monotonia</dt>
<dd>$$given \, f(x)  \le  g(x) \, \forall  \, x \,  \in [a, b] \Rightarrow \int_a^b f(x)\,dx  \le  \int_a^b g(x)\,dx$$</dd>
<dt>Acotamiento</dt>
<dd>$$ m(b-a) \le \int_a^b \, f(x) \, dx \le M(b-a)$$

Donde:

$m(b-a)$ es el valor mínimo de $f(x)$ en el rango \[a, b]

$M(b-a)$ es el valor máximo de $f(x)$ en el rango \[a, b]
</dd>
</dl>

# Valor medio de una función

Si $f(x)$ es contínua en \[a, b], su valor medio es:
$$f\_prom = (1 / (b - a)) \dot \int_a^b f(x) \, dx$$
f_prom es la altura del rectángulo con base (b - a) ciya área coincide con la del área bajo la curva en ese intervalo.
