# Sustituciones trigonometricas

| Factor a convertir | conversión | Diferencial  | Dominio de $\Theta$ | Simplificación |
|-|-|-|-|-|
| $\sqrt{a^2 - x^2}$ | $x = a·\sin{\Theta}$ | $\delta x = a·\cos{\Theta}\delta \Theta$ | $\Theta \in [-\Pi/2, \Pi/2]$ | $\sqrt{a^2 - x^2} \mapsto \sqrt{a^2 · (1 - \sin^2{\Theta})} \mapsto \sqrt{a^2 · \cos^2{\Theta}} \mapsto a·\textbar \cos{\Theta} \textbar$ |
| $\sqrt{a^2 + x^2}$ | $x = a·\tan{\Theta}$ | $\delta x = a·\sec^2{\Theta} \delta \Theta$ | $\Theta \in [-\Pi/2, \Pi/2]$ | $\sqrt{a^2 + x^2} \mapsto \sqrt{a^2 + a²·\tan²{\Theta}} \mapsto \sqrt{a^2·\sec^2{\Theta}} \mapsto a·\textbar \sec{\Theta} \textbar$ |
| $\sqrt{x^2 - a^2}$ | $x = a·\sec{\Theta}$ | $\delta x = a·\sec{\Theta}·\tan{\Theta}·\delta\Theta$| $\Theta \in [0,\Pi/2) \cup (\Pi/2,\Pi]$ | $\sqrt{x^2 - a^2} \mapsto \sqrt{a^2·\sec^2{\Theta} - a^2} \mapsto \sqrt{a^2·\tan^2{\Theta}} \mapsto a·\textbar \tan{\Theta} \textbar$ |



```python
from sympy import *
x, a = symbols('x a', positive=True)

f1 = Integer(1) / sqrt(4 - x**2)
result1 = integrate(f1, x)
print(f"Caso 1: {result1}")
f2 = Integer(1) / sqrt(1 + x**2)
result2 = integrate(f2, x)
print(f"Caso 2: {result2}")
f3 = Integer(1) / (x**2 * sqrt(x**2 + 1))
result3 = integrate(f3, x)
print(f"Caso 3: {result3}")
```
