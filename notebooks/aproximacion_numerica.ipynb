# Aproximación numérica

Reglas del trapecio y de simpson

El TFC requiere de encontrar una antiderivada, hay funciones que no tienen una antiderivada elemental, por ejemplo: $e^{-x^2}$, $sin(x)/x$ o
$\sqrt{1 + cos^2{x} }

Idea central: dividir el intervalo \[a, b] en n subintervalos y aproximar el área bajo la curva usando figuras geométricas simples (trapecios o parábolas) cuya área se podemos calcular exactamente.

## Regla del trapecio



```python
import numpy as np
from math import exp

n = 20
f = lambda x: exp( - x ** 2 )
a = -1.0
b = 1.0
dx = (b - a) / n
x = np.array([a + i * dx for i in range(0, n+1)])
y = [f(x) for x in x]
int_of_f = np.trapezoid(y, x)
print(int_of_f)
```
