# Integración por partes

$$\int u(x) dv(x) dx = u(x) · v(x) - \int v(x) du(x) dx$$

## Criterio LIATE

| Sigla | Descripción  | Ejemplos |
|-------|--------------|----------|
| L     | Logarítmicas | $\ln{x}$ |
| I     | Inversas trigonométricas | $\arctan(x)$<br/>$\arcsin(x)$<br/>$\arccos(x)$ |
| A     | Algebraicas o polinómicas | $x^{n}$<br/>`constantes` |
| T     | Trigonométricas | $\sin(x)$<br/>$\cos(x)$<br/>$\tan(x)$ |
| E     | Exponenciales | $e^{x}$<br/>$a^{x}$ |


### Ejemplo 1

Calcular la siguiente $\int{x·e^{x}}dx$
Como $x$ es A (algebraica) y $e^x$ es E (exponencial). Como A aparece antes que E en LIATE entonces seleccionamos $u = x$ y $dv = e^{x}dx$, de donde $du = dx$ y $v = e^{x}$ dando:

$$\int{x· e^{x}\,dx} = x· e^{x} - \int{e^{x}}\,dx = x· e^x - e^x + C = e^{x}(x - 1) + C$$


### Ejemplo 2

$$\int ln(x)\,dx = \int ln(x)·1\,dx$$

Usando LIATE    $u = ln(x)$ y $dv = dx$ , entonces: $du = 1/x\,dx$ y $v = x$

$$\int ln(x)\,dx = x·ln(x) - \int x·(1/x)dx = x·ln(x) - x + C$$

## Casos especiales: funciones inversas trigonométricas y productos exponenciales



```python
from sympy import symbols, exp, sin, ln, integrate, atan
x = symbols('x')
resultado1 = integrate(x * exp(x), x)
print(resultado1)
resultado2 = integrate(ln(x), x)
print(resultado2)
resultado3 = integrate(atan(x), x)
print(resultado3)
resultado4 = integrate(x**3*exp(x), x)
print(resultado4)
```
