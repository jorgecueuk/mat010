## ¿Qué es una antiderivada?

Una función F es antiderivada de f en un intervalo si F'(s) = f(x) para todo x en ese intervalo. En otras palabras, F es la función cuya derivada es f.

```python
C = 5
F = x ** 3 / 3 + C
f = sp.diff(F, x)
print(f"F(x): {F}")
print(f"f(x): {f}")
```

# Ejemplo: Integral de una función trigonométrica

```python
F = sp.sin(x)
G = sp.integrate(F, x)
print(f"F(x): {F}")
print(f"G(x): {G}")
g = sp.lambdify(x, G, "numpy")
A = g(3.141592654) -  g(0.0)
print(f"A(x): {A}")
```

```python
F = x ** 2
G = sp.integrate(F, x)
print(f"F(x): {F}")
print(f"G(x): {G}")
g = sp.lambdify(x, G, "numpy")
A = g(3) - g(0)
print(f"A(x): {A}")
```
